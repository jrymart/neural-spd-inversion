``` python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.size'] = 35           # Base font size
plt.rcParams['axes.labelsize'] = 35      # Axis labels (larger)
plt.rcParams['axes.titlesize'] = 35      # Title
plt.rcParams['xtick.labelsize'] = 25     # X tick labels
plt.rcParams['ytick.labelsize'] = 25     # Y tick labels
plt.rcParams['legend.fontsize'] = 30     # Legend
#plt.rcParams['axes.labelweight'] = 'bold'  # Bold axis labels
colors = ['#0173B2', '#DE8F05', '#029E73', '#CC78BC']  # blue, orange, teal, pink
colors = ['#60992D', '#5171A5', '#FFE787', '#D62828']
```

# Load Data

``` python
PERFORMANCE_CSV = "../../../analysis/overall_performance.csv"
PERFORMANCE_CSV = "../../../analysis/little_results.csv"
df = pd.read_csv(PERFORMANCE_CSV)
df = df[df['seed']==0]
df = df[df['noise']==0]
data_name_dict = {
    'elevation': 'Elevation',
    'slope': 'Slope',
    'curvature': 'Curvature',
    'flowacc': 'Flow Acc',
    'log10flowacc': r'$\log_{10}$'+'(Flow Acc)'
}
target_name_dict = {
    'DoK': '$D/K$',
    'KoD': '$K/D$',
    'logKoD': r'$\log_{10}(K/D)$',
    'logDoK': r'$\log_{10}(D/K)$'
}
df['Data Type'] = df['data'].map(data_name_dict)
df['Target Variable'] = df['target'].map(target_name_dict)
df.head().filter(df.columns[1:])
```

``` python
def make_plot(ax, targets = list(target_name_dict.keys())):
    fdf = df[df['target'].isin(targets)]
    g = sns.barplot(fdf, x='Data Type', y='nrmse', ax=ax, hue='Target Variable',
                order=data_name_dict.values(),
                hue_order=[target_name_dict[t] for t in targets],
                palette=colors)
#    g.set_xticklabels(g.get_xticklabels(), rotation=30)
    ax.set_ylabel("NRMSE")
```

# generate versions

``` python
fig, axs = plt.subplots(1, figsize=(15, 10) )
make_plot(axs, ["DoK"])
fig.tight_layout()
fig.savefig('DoKcomparison.png', dpi=300)
fig, axs = plt.subplots(1, figsize=(15, 10) )
make_plot(axs, ["DoK", "KoD"])
fig.tight_layout()
fig.savefig('unloggedcomparison.png', dpi=300)
fig, axs = plt.subplots(1, figsize=(15, 10) )
make_plot(axs)
fig.tight_layout()
fig.savefig('allcomparison.png', dpi=300)

```
