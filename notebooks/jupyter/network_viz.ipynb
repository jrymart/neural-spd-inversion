# Setup

``` python
import networkx as nx
from networkx.drawing.nx_agraph import graphviz_layout
import matplotlib.pyplot as plt
from neural_spd.config import PROJECT_ROOT
from matplotlib.lines import Line2D
import numpy as np
from neural_spd import plot_styles
from neural_spd.plot_styles import COLORS
plot_styles.apply()
```

## Generate network topology

``` python
nodes = np.arange(157)
links= []
for i in range(1,11):
    links.append((0,i))
    links.append((i, i+10))
    links.append((i+10, i+20))
    for j in range(1,21):
        links.append((i+20, 30+j))
for i in range(1,21):
    links.append((i+30,i+50))
    links.append((i+50,i+70))
    links.append((i+90,i+110))
    links.append((i+110,i+130))
    for j in range(1,21):
        links.append((i+70, 20+70+j))
        links.append((130+j, 151))
for i in range(5):
    links.append((151+i, 151+i+1))
edges = links
```

``` python
node_colors = [COLORS['gray_mid']]

```

# Plotting Function

``` python
def plot_network(ax):
    G = nx.DiGraph()
    G.add_nodes_from(nodes)
    G.add_edges_from(edges)
    pos = graphviz_layout(G, prog="dot")
    nx.draw(
        G, pos, ax=ax,
        with_labels=False,
        node_color=node_colors,
        edge_color=COLORS["gray_light"],
        node_size=60,
        width=0.4,
        arrowsize=4,
    )
    legend_elements = [
        Line2D([0], [0], marker="o", color="w", label="Input / Output",
               markerfacecolor=COLORS["gray_mid"], markersize=6),
        Line2D([0], [0], marker="o", color="w", label="Convolution",
               markerfacecolor=COLORS["green_mid"], markersize=6),
        Line2D([0], [0], marker="o", color="w", label="ReLU",
               markerfacecolor=COLORS["blue"], markersize=6),
        Line2D([0], [0], marker="o", color="w", label="MaxPool",
               markerfacecolor=COLORS["orange"], markersize=6),
        Line2D([0], [0], marker="o", color="w", label="Linear",
               markerfacecolor=COLORS["purple"], markersize=6),
    ]
    ax.legend(handles=legend_elements, loc="upper right", title="Node Type",
              title_fontsize=plt.rcParams["legend.fontsize"])
    ax.set_axis_off()

```

# Generate Plots

``` python
plot_styles.double_column()
fig, ax = plt.subplots(figsize=(6.75, 5.0))
plot_network(ax)
plot_styles.save_figure(fig, "network_viz", PROJECT_ROOT / "paper/figs")
```
