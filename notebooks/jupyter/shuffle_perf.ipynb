This figure shows the performance of the trained network on shuffled and

# Setup

``` python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from neural_spd.config import PROJECT_ROOT
from neural_spd import plot_styles
import numpy as np
plot_styles.apply()
reg_perf = PROJECT_ROOT / "analysis/all_test_performance.csv"
mod_perf = PROJECT_ROOT / "analysis/all_mod_performance.csv"
target_names = {
    'DoK': "$D/K$",
    'KoD': "$K/D$",
    'logDoK': r"$\log_{10}(D/K)$",
    'logKoD': r"$\log_{10}(K/D)$"
}
data_names = {
    'elevation': "Elevation",
    'flowacc': "Flow Acc",
    'log10flowacc': r"$\log_{10}(\mathrm{Flow\ Acc})$",
    'slope': "Slope",
    'curvature': "Curvature"
}
reg_df = pd.read_csv(reg_perf)
reg_df['test_type'] = 'Unmodified'
mod_df = pd.read_csv(mod_perf)
label_map = {'shuffle': 'Tile Shuffle', 'swap': 'Horizontal Swap'}
mod_df['test_type'] = mod_df['test_type'].map(label_map).fillna(mod_df['test_type'])
all_df = pd.concat([reg_df, mod_df], ignore_index=True)
elev_logDoK = all_df[
    (all_df["target"] == "logDoK") &
    (all_df["data"] == "elevation")
]
elev_logKoD = all_df[
    (all_df["target"] == "logKoD") &
    (all_df["data"] == "elevation")
]
slope_logDoK = all_df[
    (all_df["target"] == "logDoK") &
    (all_df["data"] == "slope")
]
slope_logKoD = all_df[
    (all_df["target"] == "logKoD") &
    (all_df["data"] == "slope")
]
logacc_logDoK = all_df[
    (all_df["target"] == "logDoK") &
    (all_df["data"] == "log10flowacc")
]
logacc_logKoD = all_df[
    (all_df["target"] == "logKoD") &
    (all_df["data"] == "log10flowacc")
]
curvature_logDoK = all_df[
    (all_df["target"] == "logDoK") &
    (all_df["data"] == "curvature")
]
curvature_logKoD = all_df[
    (all_df["target"] == "logKoD") &
    (all_df["data"] == "curvature")
]

```

# plotting function

``` python
_PANEL_DATA = [
    (elev_logDoK,    "Elevation",          r"$\log_{10}(D/K)$"),
    (elev_logKoD,    "Elevation",          r"$\log_{10}(K/D)$"),
    (slope_logDoK,   "Slope",              r"$\log_{10}(D/K)$"),
    (slope_logKoD,   "Slope",              r"$\log_{10}(K/D)$"),
    (logacc_logDoK,  r"$\log_{10}$ Flow Acc", r"$\log_{10}(D/K)$"),
    (logacc_logKoD,  r"$\log_{10}$ Flow Acc", r"$\log_{10}(K/D)$"),
    (curvature_logDoK, "Curvature",          r"$\log_{10}(D/K)$"),
    (curvature_logKoD, "Curvature",          r"$\log_{10}(K/D)$"),
]

def plot_shuf_perf(axs):
    for ax, (df, data_label, target_label) in zip(axs.flat, _PANEL_DATA):
        sns.kdeplot(
            data=df,
            x="true_labels",
            y="predictions",
            hue="test_type",
            fill=true,
            ax=ax,
        )
        ax.set_title(f"{data_label} — {target_label}")
        ax.set_xlabel(f"True {target_label}")
        ax.set_ylabel(f"Predicted {target_label}")
        # add 1:1 reference line
        lims = [df["true_labels"].min(), df["true_labels"].max()]
        ax.plot(lims, lims, color=plot_styles.COLORS["gray_mid"],
                linestyle="--", linewidth=0.8, zorder=3)
```

# Generate plots

``` python
plot_styles.double_column()
fig, axs = plt.subplots(3, 2, figsize=(6.75, 8.5))
plot_shuf_perf(axs)
for i, ax in enumerate(axs.flatten()):
    if i != 0:
        ax.legend().remove()
fig.suptitle("Shuffle/Swap Robustness — All Input/Target Combinations",
             fontsize=plt.rcParams["axes.titlesize"], y=1.01)
plot_styles.save_figure(fig, "all_shuffle_perf", directory=PROJECT_ROOT / "paper/figs")
```
