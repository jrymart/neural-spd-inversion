# Load data

``` python
perf_metrics_df = pd.read_csv(PERF_METRICS_PATH)
perf_metrics_df["target"] = perf_metrics_df["target"].map(target_names)
perf_metrics_df["data"] = perf_metrics_df["data"].map(data_names)
swap_metrics = perf_metrics_df[perf_metrics_df['test_type'] == 'swap']
shuf_metrics = perf_metrics_df[perf_metrics_df['test_type'] == 'shuffle']
```

# Plotting function

``` python
def plot_perf_comp(axs):
   sns.barplot(data=shuf_metrics, x="data", y="nrmse", hue="target", ax=axs[0])
   axs[0].set_ylabel("NRMSE")
   axs[0].set_xlabel("Data Type")
   axs[0].set_title("Shuffled Data")
   sns.barplot(data=swap_metrics, x="data", y="nrmse", hue="target", ax=axs[1])
   #axs[0].set_ylabel("NRMSE")
   axs[1].set_xlabel("Data Type")
   axs[1].set_title("Swapped Data")
```

# Generate plots

``` python
fig, axs = plt.subplots(2, figsize=(6, 4))
plot_perf_comp(axs)
fig.show()
```
