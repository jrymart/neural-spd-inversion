# Overview

This figure shows the performance of the elevation and slope networks on
shuffled and swapped data. :scope: paper

# Setup

``` python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from neural_spd.config import PROJECT_ROOT
from neural_spd import plot_styles
from neural_spd.plot_styles import cm
import numpy as np
plot_styles.apply()
reg_perf = PROJECT_ROOT / "analysis/all_test_performance.csv"
mod_perf = PROJECT_ROOT / "analysis/all_mod_performance.csv"

reg_df = pd.read_csv(reg_perf)
reg_df['test_type'] = 'Unmodified'
mod_df = pd.read_csv(mod_perf)
label_map = {'shuffle': 'Tile Shuffle', 'swap': 'Horizontal Swap'}
mod_df['test_type'] = mod_df['test_type'].map(label_map).fillna(mod_df['test_type'])
all_df = pd.concat([reg_df, mod_df], ignore_index=True)
elev_logDoK = all_df[
    (all_df["target"] == "logDoK") &
    (all_df["data"] == "elevation")
]
slope_logDoK = all_df[
    (all_df["target"] == "logDoK") &
    (all_df["data"] == "slope")
]
```

# Plotting Function

``` python
def plot_shuf_perf(axs):
    sns.scatterplot(
            data=elev_logDoK,
            x="true_labels",
            y="predictions",
            hue="test_type",
            s=2,
            edgecolor=None,
            #fill=True,
            ax=axs[0],
        )
    #axs[0].set_title(r"Elevation — $\log_{10}(D/K)$")
    axs[0].text(0.4, 2, "(a)")
    axs[0].set_xlabel(r"True $\log_{10}(D/K)$")
    axs[0].set_ylabel(r"Predicted $\log_{10}(D/K)$")
    # add 1:1 reference line
    lims = [elev_logDoK["true_labels"].min(), elev_logDoK["true_labels"].max()]
    axs[0].plot(lims, lims, color=plot_styles.COLORS["gray_mid"],
            linestyle="--", linewidth=0.8, zorder=3)
    sns.kdeplot(
            data=slope_logDoK,
            x="true_labels",
            y="predictions",
            hue="test_type",
            fill=True,
            ax=axs[1],
        )
#    axs[1].set_title(r"Slope — $\log_{10}(D/K)$")
    axs[1].text(0.3, 2.3, "(b)")
    axs[1].set_xlabel(r"True $\log_{10}(D/K)$")
    axs[1].set_ylabel(r"Predicted $\log_{10}(D/K)$")
    # add 1:1 reference line
    lims = [slope_logDoK["true_labels"].min(), slope_logDoK["true_labels"].max()]
    axs[1].plot(lims, lims, color=plot_styles.COLORS["gray_mid"],
            linestyle="--", linewidth=0.8, zorder=3)
    axs[1].legend().remove()

```

# Generate Plots

``` python
plot_styles.double_column()
fig, axs = plt.subplots(1,2, figsize=(17*cm, 8.5*cm))
plot_shuf_perf(axs)
plot_styles.save_figure(fig, "shuff_perf_paper", directory=PROJECT_ROOT / "paper/figs")
fig.show()
```
