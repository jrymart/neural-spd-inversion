# Overview

This figure shows the performance of a network on the sine dataset.

analysis/sine<sub>allperformance</sub>.csv

# Setup

``` python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from neural_spd.config import PROJECT_ROOT
from neural_spd import plot_styles
from itertools import product
plot_styles.apply()
sine_perf = PROJECT_ROOT / "analysis/sine_all_performance.csv"
sine_df = pd.read_csv(sine_perf)
sine_df = sine_df[sine_df["target"] == "logDoK"]
sine_df = sine_df[sine_df['seed']==20]
sine_df = sine_df[sine_df['data'] =="elevation"]
sine_df = sine_df[sine_df['noise']==0.0]

```

# Plotting function

``` python
def plot_perf(ax):
        sns.scatterplot(
            data=sine_df,
            x="true_labels",
            y="predictions",
            s=2,
            ax=ax,
            palette="Set2"
        )
        ax.set_xlabel("Valley Spacing (m)")
        ax.set_ylabel(r"$\log_{10}(D/K)$")
```

# Generate plots

``` python
fig, ax = plt.subplots(figsize=(3.25,3.25))
plot_perf(ax)
fig.show()
```
